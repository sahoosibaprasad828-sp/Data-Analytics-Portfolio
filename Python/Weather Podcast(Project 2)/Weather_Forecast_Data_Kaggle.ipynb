```python
import numpy as np
import pandas as pd
df = pd.read_csv("weather_forecast_data.csv")
df
```

```python
df.info()
```

```python
# Check for missing values
print(df.isnull().sum())

# Get statistical summary of numerical columns
df.describe()
```

```python
# Group by 'Rain' and find the average of weather conditions
df.groupby('Rain').mean()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for cleaner-looking charts
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='YlGnBu', fmt=".2f")
plt.title("Correlation Matrix of Weather Variables")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Temperature'], kde=True, color='orange')
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='Rain', y='Humidity', data=df, hue='Rain', palette='Set2', legend=False)
plt.title("Humidity Levels on Rainy vs. Non-Rainy Days")
plt.show()
```

## Project Summary: Weather Forecast Data Analysis

### 1. Project Overview
This project focuses on performing **Exploratory Data Analysis (EDA)** on a weather dataset using Python within a Jupyter Notebook environment. The primary objective is to inspect weather variables, uncover patterns, and visualize how environmental factors correlate with weather conditions like rain.

### 2. Tools & Libraries Used
* **Python** (Programming Language)
* **Pandas** (Data manipulation and loading)
* **Matplotlib & Seaborn** (Data visualization and statistical plotting)

### 3. Key Steps & Analysis Performed
* **Data Loading & Inspection:** Loaded the weather forecast dataset to inspect its structure, rows, columns, and data types.
* **Univariate Analysis (Temperature Distribution):** Generated a histogram with a KDE curve using Seaborn to analyze the spread, frequency, and central tendencies of temperatures.
* **Bivariate Analysis (Humidity vs. Rain):** Created a box plot comparing humidity levels on rainy versus non-rainy days. 

### 4. Major Insights Discovered
* **Humidity as a Key Predictor:** The analysis clearly demonstrated that **rainy days feature significantly higher humidity levels** compared to non-rainy days, establishing humidity as a core indicator for precipitation.
* **Temperature Trends:** The distribution plots mapped out the frequency of common temperature ranges, making it easy to identify standard weather conditions.
