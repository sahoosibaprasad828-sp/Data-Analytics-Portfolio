```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Set style for cleaner graphs
sns.set_theme(style="whitegrid")
```

```python
print(df['gender'].value_counts())
print(df['Investment_Avenues'].value_counts())
```

```python
avg_age_by_avenue = (
    df.groupby("Investment_Avenues")["age"].mean().reset_index()
)
print(avg_age_by_avenue)
```

```python
gender_avenue_cross = pd.crosstab(df["gender"], df["Avenue"])
print(gender_avenue_cross)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["age"], bins=10, kde=True, color="teal")
plt.title("Age Distribution of Investors", fontsize=14, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x="Avenue", hue="gender", palette="Set2")
plt.title(
    "Preferred Investment Avenue by Gender", fontsize=14, fontweight="bold"
)
plt.xlabel("Investment Avenue")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 8))
# Selecting only numerical rating columns
rating_cols = [
    "Mutual_Funds",
    "Equity_Market",
    "Debentures",
    "Government_Bonds",
    "Fixed_Deposits",
    "PPF",
    "Gold",
]
correlation_matrix = df[rating_cols].corr()

sns.heatmap(
    correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5
)
plt.title(
    "Correlation Heatmap of Investment Preferences",
    fontsize=14,
    fontweight="bold",
)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
rating_cols = [
    "Mutual_Funds",
    "Equity_Market",
    "Debentures",
    "Government_Bonds",
    "Fixed_Deposits",
    "PPF",
    "Gold",
]
mean_scores = df[rating_cols].mean().sort_values(ascending=False)

sns.barplot(x=mean_scores.index, y=mean_scores.values, palette="viridis")
plt.title(
    "Average Preference Ranking Across Investment Avenues",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Investment Avenue", fontsize=12)
plt.ylabel("Average Score", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

```python
# Create a Pie Chart for Average Investment Preferences
plt.figure(figsize=(8, 8))
rating_cols = [
    "Mutual_Funds",
    "Equity_Market",
    "Debentures",
    "Government_Bonds",
    "Fixed_Deposits",
    "PPF",
    "Gold",
]
mean_scores = df[rating_cols].mean().sort_values(ascending=False)

plt.pie(
    mean_scores.values,
    labels=mean_scores.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=sns.color_palette("pastel"),
    wedgeprops={"edgecolor": "black", "linewidth": 1},
)

plt.title(
    "Share of Average Preference Ranking Across Investment Avenues",
    fontsize=14,
    fontweight="bold",
)
plt.tight_layout()
plt.show()
```
